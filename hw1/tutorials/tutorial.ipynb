```python
import sys
sys.path.append("..")
from vectorlib import Vector
```

```python

a = Vector([1, 2, 3, 5, 6])
b = Vector([10, 20, 30, 40,50])

print(a + b)     # сложение
print(b - a)     # вычитание
print(a * b)     # умножение поэлементно
print(b / a)     # деление поэлементно
print(a @ b)     # скалярное произведение (число)
print(a[0])      # индекс
print(a[1:4])    # срез
```
